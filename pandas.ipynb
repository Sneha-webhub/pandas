```python
import numpy as np
import pandas as pd

# Creating a DataFrame from scratch
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 28],
    "City": ["New York", "London", "Paris", "London", "Tokyo"],
    "Salary": [70000, 80000, 120000, 110000, 95000],
}

data = pd.DataFrame(data)
print(data)
# to edit index
data.index = [101,102,103,104,105]
print(data)
# To make any column as index
data.set_index('Name', inplace=True)
print(data)

# Data Inspection
print("=== Initial DataFrame ===")
print(data)
print()
print("=== Summary Info ===")
print(data.info())
print("=== First 5 rows ===")
print(data.head())
print("=== Lats 3 rows ===")
print(data.tail(4))
print("=== shape ===")
print(data.shape)
print("=== Column name ===")
print(data.columns)
print("=== Descriptive Statistics ===")
print(data.describe())
```
